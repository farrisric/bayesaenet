```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


plt.rcParams.update({
    "font.family": "sans-serif",
    "font.sans-serif": "Arial",
    "font.size" : 50,
    'mathtext.default':  'regular'
        })
```

```python
# Load the data
file_path = 'uq_metrics_Test.csv'
sns.set_context("talk", font_scale=2)
data = pd.read_csv(file_path)

melted = pd.melt(data, id_vars=['Method', 'Size'], value_vars=['mae', 'rmse', 'nll'],
                 var_name='Metric', value_name='Value')
melted['Method'] = melted['Method'].str.upper()                
# Set visual style and context for better aesthetics
sns.set(style="whitegrid", context="talk")

# Create box plots with improved styling
g = sns.catplot(
    data=melted,
    kind='box',
    x='Size',
    y='Value',
    hue='Method',
    col='Metric',
    palette='Set2',
    height=5,
    aspect=1.1,
    sharey=False,
    linewidth=1.2,
    dodge=True,
    col_wrap=2, 
)

# Customize titles and labels
for ax, title in zip(g.axes.flat, ['MAE', 'RMSE', 'NLL']):
    ax.set_title(title)
    ax.set_xlabel("")

g.set_axis_labels("Data Regime", "Metric Value")
g._legend.set_title("Method")

g._legend.set_bbox_to_anchor((0.5, 0.35))
g._legend.set_loc('center left')

g.savefig('performace_de_bnn.png')
```
